```python
import numpy as np
```

```python
T = 1.0
F = 0.0
```

```python
def getOrData():
  X = [
      [F,F],
      [T,F],
      [F,T],
      [T,T]
  ]

  Y = [
      [F],
      [T],
      [T],
      [T]
  ]

  return X,Y

X,Y = getOrData()
```

```python
print(X)
print(Y)
```

```python
class MyPerceptron:
    def __init__(self, learning_rate=0.1, n_iterations=1000):
        self.lr = learning_rate
        self.epochs = n_iterations
        self.weights = None
        self.bias = None

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def Step_activ_func(self, activation):
        return 1 if activation >= 0 else 0

    def relu(self,z):
      return max(0,z)

    def fit(self, X, Y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for epoch in range(self.epochs):
            for i in range(X.shape[0]):
                Y_pred = self.relu(np.dot(self.weights, X[i]) + self.bias)
                error = Y[i] - Y_pred
                self.weights += self.lr * error * X[i]
                self.bias += self.lr * error

    def predict(self, X):
        Y_pred = []
        for i in range(X.shape[0]):
            y = self.Step_activ_func(np.dot(self.weights, X[i]) + self.bias)
            Y_pred.append(y)
        return np.array(Y_pred)
```

```python
clf = MyPerceptron()

X = np.array(X)
Y = np.array(Y)
clf.fit(X,Y)

X_test = [
    [F,F],
    [T,F],
    [F,T],
    [F,F]
]

Y_test = [
    [F],
    [T],
    [T],
    [F]
]

X_test = np.array(X_test)
Y_test = np.array(Y_test)
Y_predicted = clf.predict(X_test)
print(Y_predicted)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test, Y_predicted)
print(accuracy)
```
